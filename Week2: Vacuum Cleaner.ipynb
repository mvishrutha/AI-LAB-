```python
room1 = "Dirty"
room2 = "Clean"

vacuum = "Room 1"

if vacuum == "Room 1":
    if room1 == "Dirty":
        print("Room 1 is dirty. Cleaning...")
        room1 = "Clean"

    print("Room 1 is clean. Moving to Room 2.")
    vacuum = "Room 2"

if vacuum == "Room 2":
    if room2 == "Dirty":
        print("Room 2 is dirty. Cleaning...")
        room2 = "Clean"
    else:
        print("Room 2 is already clean.")

print("Room 1:", room1)
print("Room 2:", room2)
```
