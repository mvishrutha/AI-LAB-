```python
import random

def calculate_heuristic(state):
    """
    Calculates the number of mutually attacking pairs of queens.
    This is our heuristic h(s) which we want to minimize to 0.
    """
    n = len(state)
    attacks = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Check if they are in the same row or the same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_neighbors(state):
    """
    Generates all possible neighboring states by moving any single queen
    to any other row in its column. For N columns, there are N * (N - 1) neighbors.
    """
    n = len(state)
    neighbors = []
    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbors.append(tuple(neighbor))
    return neighbors

def hill_climbing_from_initial(initial_state, max_restarts=100):
    """
    Solves the N-Queens problem using Hill Climbing.
    Starts with the user's initial configuration. If it gets stuck,
    it uses random restarts to find the global minimum (solution).
    """
    n = len(initial_state)
    current_state = tuple(initial_state)
    current_h = calculate_heuristic(current_state)

    print(f"\n--- Start of Hill Climbing (Using Initial State) ---")
    print(f"Initial State: {current_state} with h = {current_h} attacking pairs")

    step = 0
    while True:
        if current_h == 0:
            print(f"\nSuccess! Solved from initial state in {step} steps.")
            return current_state

        neighbors = get_neighbors(current_state)
        neighbor_heuristics = [(neighbor, calculate_heuristic(neighbor)) for neighbor in neighbors]
        best_neighbor, best_h = min(neighbor_heuristics, key=lambda x: x[1])

        if best_h >= current_h:
            print(f"Stuck in a local minimum at step {step} with h = {current_h}.")
            print("Switching to Random Restarts to find a solution...")
            break

        current_state = best_neighbor
        current_h = best_h
        step += 1
        print(f"Step {step}: State {current_state} | h = {current_h}")

    # If the custom state got stuck, fallback to random restarts to solve it
    for restart in range(max_restarts):
        current_state = tuple(random.randint(0, n - 1) for _ in range(n))
        current_h = calculate_heuristic(current_state)

        print(f"\n--- Start of Restart {restart + 1} ---")
        print(f"Initial State: {current_state} with h = {current_h} attacking pairs")

        step = 0
        while True:
            if current_h == 0:
                print(f"\nSuccess! Solved in restart {restart + 1}, step {step}.")
                return current_state

            neighbors = get_neighbors(current_state)
            neighbor_heuristics = [(neighbor, calculate_heuristic(neighbor)) for neighbor in neighbors]
            best_neighbor, best_h = min(neighbor_heuristics, key=lambda x: x[1])

            if best_h >= current_h:
                print(f"Stuck in a local minimum at step {step} with h = {current_h}. Restarting...")
                break

            current_state = best_neighbor
            current_h = best_h
            step += 1
            print(f"Step {step}: State {current_state} | h = {current_h}")

    print("\nCould not find a solution within the maximum number of restarts.")
    return None

def print_board(state):
    """
    Prints the board state visually.
    """
    n = len(state)
    for row in range(n):
        line = []
        for col in range(n):
            if state[col] == row:
                line.append("Q ")
            else:
                line.append(". ")
        print("".join(line))
    print()

# Get user input for N (number of queens / rows)
while True:
    try:
        N = int(input("Enter the number of queens (N): "))
        if N >= 4:
            break
        else:
            print("N must be 4 or greater for a standard N-Queens solution.")
    except ValueError:
        print("Please enter a valid integer.")

# Get custom initial state from user
initial_state = []
print(f"\nEnter the row index (0 to {N-1}) for each column to set your initial state:")
for col in range(N):
    while True:
        try:
            row = int(input(f"  Row for Column {col} (c{col}): "))
            if 0 <= row < N:
                initial_state.append(row)
                break
            else:
                print(f"Please enter a row index between 0 and {N-1}.")
        except ValueError:
            print("Please enter a valid integer.")

initial_state = tuple(initial_state)

# Solve the problem starting with user configuration
solution = hill_climbing_from_initial(initial_state)
if solution:
    print(f"Final Solution State: {solution}")
    print("\nVisual representation of the board:")
    print_board(solution)
```
